```python
"""
Q1 – Pass or Fail Counter

Write a script that:
1. Loops through the marks and prints Pass or Fail.
2. Counts passed and failed marks.
3. Creates a list of passing marks using comprehension.
4. Prints the final counts and passing marks.
"""

marks = [45, 32, 78, 39, 90, 28, 55, 40, 67, 12]

passed = 0
failed = 0

for mark in marks:
    if mark >= 40:
        print(mark, "Pass")
        passed += 1
    else:
        print(mark, "Fail")
        failed += 1

passing_marks = [mark for mark in marks if mark >= 40]

print("Passed:", passed)
print("Failed:", failed)
print("Passing Marks:", passing_marks)
```

```python
'''
4  Question Student class
'''

class student:
  def __init__(self,name: str,marks: list):

        self.name = name
        self.marks = marks
  def __str__(self) ->str:
    avg = sum(self.marks)/len(self.marks)
    return f"{self.name} : average{avg}"


student1 = student("Sakshi",[75,80,72,78])
Student2 = student("Mayur", [90, 88, 92])

print(student1)
print(Student2)
```

```python
"""
Q3 – Daily Step Tracker
"""

import random
import math

steps = []

for day in range(1, 8):
    count = random.randint(2000, 12000)
    steps.append(count)
    print(f"Day {day}: {count} steps")

print("Highest:", max(steps))
print("Lowest:", min(steps))

average = sum(steps) / len(steps)
print("Average:", math.ceil(average))
```

```python
"""
Q2 – Safe Average Calculator
"""

def calculate_average(numbers: list) -> float:
    """Returns the average of a list of numbers."""
    return sum(numbers) / len(numbers)


try:
    print("Average:", calculate_average([10, 20, 30]))
except ZeroDivisionError:
    print("Cannot calculate average of an empty list.")

try:
    print("Average:", calculate_average([]))
except ZeroDivisionError:
    print("Cannot calculate average of an empty list.")
```
